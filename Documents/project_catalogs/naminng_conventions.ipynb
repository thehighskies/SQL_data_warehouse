## 2. Core naming rules

### Rule 1: Use lowercase for all SQL object names
All object names in this project are lowercase.

Examples:
- `bronze.crm_cust_info`
- `silver.erp_loc_a101`
- `gold.dim_customers`
- `bronze.load_bronze`

### Rule 2: Always include the schema prefix
Every object must be referenced as `schema.object_name`.

Examples:
- `bronze.crm_cust_info`
- `silver.crm_sales_details`
- `gold.fact_sales`

### Rule 3: Use the same source/entity pattern consistently
The project follows the pattern:

- `crm_...` for CRM source objects
- `erp_...` for ERP source objects
- `cust`, `prd`, `sales`, `loc`, `cat` for common domain terms

Examples:
- `crm_cust_info`
- `crm_prd_info`
- `crm_sales_details`
- `erp_loc_a101`
- `erp_px_cat_g1v2`

### Rule 4: Keep abbreviation style consistent with the current repo
The project uses these abbreviations as the standard style:

| Term | Standard abbreviation used in project |
|---|---|
| customer | `cust` |
| product | `prd` |
| sales | `sls` |
| location | `loc` |
| category | `cat` |
| information | `info` |
| details | `details` |
| CRM | `crm` |
| ERP | `erp` |
| dimension | `dim` |
| fact | `fact` |
| load | `load` |

These are the repo-approved forms. Do not replace them with alternate spellings such as `customer`, `product_detail`, or `tbl_*`.

## 3. Naming by layer

### 3.1 Bronze layer naming
Bronze tables are raw source tables and follow this pattern:

`{source}_{entity}_{descriptor}`

Examples from the repo:
- `bronze.crm_cust_info`
- `bronze.crm_prd_info`
- `bronze.crm_sales_details`
- `bronze.erp_cust_az12`
- `bronze.erp_loc_a101`
- `bronze.erp_px_cat_g1v2`

Notes:
- The source prefix is always `crm` or `erp`.
- The entity and descriptor names follow the repository's short naming style.
- Do not invent a different bronze naming format.

### 3.2 Silver layer naming
Silver tables follow the same naming pattern as bronze tables.

Pattern:

`silver.{source}_{entity}_{descriptor}`

Examples:
- `silver.crm_cust_info`
- `silver.crm_prd_info`
- `silver.crm_sales_details`
- `silver.erp_cust_az12`
- `silver.erp_loc_a101`
- `silver.erp_px_cat_g1v2`

The silver layer is not a different naming family; it mirrors the bronze layer naming convention after cleaning and transformation.

### 3.3 Gold layer naming
Gold objects follow dimensional modeling conventions already present in the project.

Pattern:

- Dimensions: `gold.dim_{entity}`
- Facts: `gold.fact_{entity}`

Actual project examples:
- `gold.dim_customers`
- `gold.dim_products`
- `gold.fact_sales`

Do not create a different naming convention like `gold.customer_dim` or `gold.sales_fact_table` unless the project intentionally adopts that later.

### 3.4 Stored procedure naming
Stored procedures follow the actual pattern already used in the project:

`{schema}.load_{target}`

Actual project examples:
- `bronze.load_bronze`
- `silver.load_silver`

This is the correct convention for this repository. The verbs and target names are not arbitrary; they reflect the current implementation.

## 6. Naming rules to follow without exception

1. Use lowercase only.
2. Use schema-qualified names: `schema.object_name`.
3. Keep the repository abbreviations exactly as used: `crm`, `erp`, `cust`, `prd`, `sls`, `loc`, `cat`, `info`, `dim`, `fact`.
4. Use the bronze/silver naming pattern for source tables.
5. Use the gold pattern `dim_{entity}` and `fact_{entity}` for curated objects.
6. Use the procedure pattern `{schema}.load_{target}`.
7. Do not invent new naming variants that are not present in the existing project.

---

## 7. Final project convention summary

The approved naming convention for this project is:

- Bronze tables: `bronze.{source}_{entity}_{descriptor}`
- Silver tables: `silver.{source}_{entity}_{descriptor}`
- Gold views: `gold.dim_{entity}` and `gold.fact_{entity}`
- Stored procedures: `{schema}.load_{target}`
- Standard abbreviations: `crm`, `erp`, `cust`, `prd`, `sls`, `loc`, `cat`, `info`, `dim`, `fact`

This is the naming standard that matches the repository as it currently exists.
